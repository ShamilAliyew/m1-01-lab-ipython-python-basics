Aliasing: If you assign one variable to another (e.g., alias = values), both names point to the same object. Changing it through one name also changes the other.

Copying: If you make a copy (e.g., values.copy()), it creates a new object. Changes to the copy do not affect the original.

For dictionaries: The same rules apply. An alias shares the object, a copy creates a new one.

Why it matters: When passing mutable objects into functions, using an alias can modify the original object, but a copy keeps it safe.

```python
# Task 4
events = [
    {"user_id": 1, "event_type": "login", "duration_seconds": 120},
    {"user_id": 2, "event_type": "logout", "duration_seconds": 300},
    {"user_id": 3, "event_type": "purchase", "duration_seconds": 180},
    {"user_id": 4, "event_type": "login", "duration_seconds": -50},
    {"user_id": 5, "event_type": "logout", "duration_seconds": 90},
    {"user_id": 6, "event_type": "purchase", "duration_seconds": "abc"},
    {"user_id": 7, "event_type": "login", "duration_seconds": 240},
    {"user_id": 8, "event_type": "logout", "duration_seconds": 60},
    {"user_id": 9, "event_type": "purchase", "duration_seconds": 150},
    {"user_id": 10, "event_type": "login", "duration_seconds": 200},
    {"user_id": 11, "event_type": "logout", "duration_seconds": 180},
    {"user_id": 12, "event_type": "purchase", "duration_seconds": 210}
]

cleaned_events = []

for event in events:
    duration = event.get("duration_seconds")
    
    try:
        duration = float(duration)
        if duration > 0:
            
            new_event = event.copy()
            new_event["duration_seconds"] = duration
            new_event["duration_minutes"] = duration / 60
            cleaned_events.append(new_event)
    except (ValueError, TypeError):
        pass  


print("Number of cleaned events:", len(cleaned_events))

print("Keys in first cleaned record:", cleaned_events[0].keys())


print("First cleaned record:", cleaned_events[0])
print("Second cleaned record:", cleaned_events[1])
```

```python
# Task 5
event_counts = {}
for event in cleaned_events:
    etype = event["event_type"]
    event_counts[etype] = event_counts.get(etype, 0) + 1

duration_totals = {}
duration_counts = {}

for event in cleaned_events:
    etype = event["event_type"]
    duration = event["duration_minutes"]
    duration_totals[etype] = duration_totals.get(etype, 0) + duration
    duration_counts[etype] = duration_counts.get(etype, 0) + 1

avg_duration = {etype: duration_totals[etype]/duration_counts[etype] for etype in duration_totals}


unique_users = set()
for event in cleaned_events:
    unique_users.add(event["user_id"])
unique_user_count = len(unique_users)

assert sum(event_counts.values()) == len(cleaned_events), "Event count mismatch!"

csv_lines = ["metric,key,value"]

for etype, count in event_counts.items():
    csv_lines.append(f"count,{etype},{count}")

for etype, avg in avg_duration.items():
    csv_lines.append(f"average_duration,{etype},{avg:.2f}")


csv_lines.append(f"unique_users,,{unique_user_count}")


csv_string = "\n".join(csv_lines)

print(csv_string)
```
